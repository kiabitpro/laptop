```python
# Prepare the CPU column
cpu = (
    df["Cpu"]
    .fillna("")
    .astype(str)
    .str.strip()
)


# Define a function to extract CPU information
def extract_cpu_info(cpu_text):

    # Create a dictionary for the extracted CPU features
    result = {
        "Cpu_Brand": np.nan,
        "Cpu_Type": np.nan,
        "processor_power": np.nan,
        "Cpu_Model": np.nan,
        "Cpu_GHz": np.nan
    }

    # Return empty values if the CPU text is empty
    if not cpu_text:
        return pd.Series(result)

    # Remove leading and trailing spaces
    text = cpu_text.strip()


    # Detect the CPU brand
    if re.search(r"\bIntel\b", text, re.IGNORECASE):
        result["Cpu_Brand"] = "Intel"

    elif re.search(r"\bAMD\b", text, re.IGNORECASE):
        result["Cpu_Brand"] = "AMD"

    elif re.search(r"\bSamsung\b", text, re.IGNORECASE):
        result["Cpu_Brand"] = "Samsung"


    # Define possible CPU types
    cpu_types = [
        "Core",
        "Celeron",
        "Pentium",
        "Atom",
        "Xeon",
        "Ryzen",
        "A-Series",
        "E-Series",
        "FX",
        "Cortex"
    ]


    # Search for the CPU type
    for cpu_type in cpu_types:

        if re.search(
            rf"\b{re.escape(cpu_type)}\b",
            text,
            re.IGNORECASE
        ):
            result["Cpu_Type"] = cpu_type
            break


    # Extract Intel Core processor power such as i3, i5, i7, and i9
    match = re.search(
        r"\b(i[3579])\b",
        text,
        re.IGNORECASE
    )

    if match:
        result["processor_power"] = match.group(1)


    # Extract Intel Core M processor power
    if pd.isna(result["processor_power"]):

        match = re.search(
            r"\b(M[357]?)\b",
            text,
            re.IGNORECASE
        )

        if match:
            result["processor_power"] = match.group(1)


    # Extract AMD A-Series processor power
    if pd.isna(result["processor_power"]):

        match = re.search(
            r"\b(A[4-9]|A1[0-2])(?:-Series)?\b",
            text,
            re.IGNORECASE
        )

        if match:
            result["processor_power"] = match.group(0)


    # Extract AMD E-Series processor power
    if pd.isna(result["processor_power"]):

        match = re.search(
            r"\b(E[1-9])\b",
            text,
            re.IGNORECASE
        )

        if match:
            result["processor_power"] = match.group(1)


    # Search for the CPU frequency in GHz
    ghz_match = re.search(
        r"(\d+(?:\.\d+)?)\s*GHz",
        text,
        re.IGNORECASE
    )


    # Continue if a GHz value was found
    if ghz_match:

        # Convert the CPU frequency to a float
        result["Cpu_GHz"] = float(
            ghz_match.group(1)
        )

        # Get everything before the GHz value
        before_ghz = text[:ghz_match.start()].strip()


        # Remove the CPU brand from the beginning
        before_ghz = re.sub(
            r"^(Intel|AMD|Samsung)\s+",
            "",
            before_ghz,
            flags=re.IGNORECASE
        ).strip()


        # Split the CPU text into separate tokens
        tokens = before_ghz.split()

        # Get the processor power
        power = result["processor_power"]


        # Continue if there are tokens
        if tokens:

            # Check if processor power was found
            if not pd.isna(power):

                # Initialize the processor power index
                power_index = None


                # Find the position of the processor power
                for i, token in enumerate(tokens):

                    if token.lower() == str(power).lower():
                        power_index = i
                        break


                # Extract the model after the processor power
                if power_index is not None:

                    remaining = tokens[
                        power_index + 1:
                    ]


                    # Check if a model exists
                    if remaining:

                        # Take the first token as the CPU model
                        model = remaining[0]


                        # Remove unwanted characters
                        model = re.sub(
                            r"[^A-Za-z0-9\-]+",
                            "",
                            model
                        )


                        # Save the CPU model
                        if model:
                            result["Cpu_Model"] = model


            # Extract the model when processor power is not available
            else:

                # Take the last token before GHz
                model = tokens[-1]


                # Remove unwanted characters
                model = re.sub(
                    r"[^A-Za-z0-9\-]+",
                    "",
                    model
                )


                # Save the CPU model
                if model:
                    result["Cpu_Model"] = model


    # Return the extracted CPU information
    return pd.Series(result)


# Apply the extraction function to the CPU column
cpu_features = cpu.apply(
    extract_cpu_info
)


# Define the new CPU columns
cpu_columns = [
    "Cpu_Type",
    "processor_power",
    "Cpu_Model",
    "Cpu_GHz"
]


# Add the extracted CPU columns to the DataFrame
df[cpu_columns] = cpu_features[cpu_columns]


# Remove the Cpu_Brand column
if "Cpu_Brand" in df.columns:
    df.drop(
        columns=["Cpu_Brand"],
        inplace=True
    )


# Get the current column order
columns = df.columns.tolist()


# Remove the CPU feature columns from their current positions
for column in cpu_columns:
    columns.remove(column)


# Insert Cpu_Type at index 8
columns.insert(8, "Cpu_Type")

# Insert processor_power at index 9
columns.insert(9, "processor_power")

# Insert Cpu_Model at index 10
columns.insert(10, "Cpu_Model")

# Insert Cpu_GHz at index 11
columns.insert(11, "Cpu_GHz")


# Apply the new column order
df = df[columns]


# Display the first 20 rows
print("\nCPU Information:")

print(
    df[
        [
            "Cpu",
            "Cpu_Type",
            "processor_power",
            "Cpu_Model",
            "Cpu_GHz"
        ]
    ].head(20)
)


# Check missing values
print("\nMissing Values:")

print(
    df[
        [
            "Cpu_Type",
            "processor_power",
            "Cpu_Model",
            "Cpu_GHz"
        ]
    ].isnull().sum()
)


# Display CPUs where the model was not extracted
print("\nCPU Model Missing:")

print(
    df[
        df["Cpu_Model"].isnull()
    ][
        [
            "Cpu",
            "Cpu_Type",
            "processor_power",
            "Cpu_Model",
            "Cpu_GHz"
        ]
    ].head(30)
)


# Display unique CPU types
print("\nUnique Values - CPU Type:")

print(
    df["Cpu_Type"]
    .value_counts(dropna=False)
)


# Display unique processor power values
print("\nUnique Values - Processor Power:")

print(
    df["processor_power"]
    .value_counts(dropna=False)
)


# Display unique CPU models
print("\nUnique Values - CPU Model:")

print(
    df["Cpu_Model"]
    .value_counts(dropna=False)
)


# Display 30 random CPU samples
print("\nRandom CPU Samples:")

print(
    df[
        [
            "Cpu",
            "Cpu_Type",
            "processor_power",
            "Cpu_Model",
            "Cpu_GHz"
        ]
    ]
    .sample(30, random_state=42)
    .to_string(index=False)
)


# Display CPUs where the type was not extracted
print("\nCPU Type Missing:")

print(
    df[
        df["Cpu_Type"].isna()
    ]["Cpu"]
    .unique()
)
```

```python
# Prepare the GPU column
gpu = (
    df["Gpu"]
    .fillna("")
    .astype(str)
    .str.strip()
)


# Define a function to extract GPU information
def extract_gpu_info(gpu_text):

    # Create a dictionary for the extracted GPU features
    result = {
        "Gpu_Type": np.nan,
        "Gpu_Model": np.nan
    }

    # Return empty values if the GPU text is empty
    if not gpu_text:
        return pd.Series(result)

    # Remove leading and trailing spaces
    text = gpu_text.strip()


    # Define possible GPU types
    gpu_types = [
        "HD Graphics",
        "UHD Graphics",
        "Iris Graphics",
        "Iris Plus Graphics",
        "GeForce",
        "Quadro",
        "Radeon",
        "Radeon Pro",
        "Mali"
    ]


    # Search for the GPU type
    for gpu_type in gpu_types:

        if re.search(
            rf"\b{re.escape(gpu_type)}\b",
            text,
            re.IGNORECASE
        ):
            result["Gpu_Type"] = gpu_type
            break


    # Search for Intel HD Graphics model
    match = re.search(
        r"(?:HD Graphics|UHD Graphics|Iris Graphics|Iris Plus Graphics)\s+(\d+)",
        text,
        re.IGNORECASE
    )

    if match:
        result["Gpu_Model"] = match.group(1)


    # Search for Nvidia GeForce model
    if pd.isna(result["Gpu_Model"]):

        match = re.search(
            r"\bGeForce\s+(.+)$",
            text,
            re.IGNORECASE
        )

        if match:
            model = match.group(1).strip()

            result["Gpu_Model"] = model


    # Search for Nvidia Quadro model
    if pd.isna(result["Gpu_Model"]):

        match = re.search(
            r"\bQuadro\s+(.+)$",
            text,
            re.IGNORECASE
        )

        if match:
            model = match.group(1).strip()

            result["Gpu_Model"] = model


    # Search for AMD Radeon model
    if pd.isna(result["Gpu_Model"]):

        match = re.search(
            r"\bRadeon(?:\s+Pro)?\s+(.+)$",
            text,
            re.IGNORECASE
        )

        if match:
            model = match.group(1).strip()

            result["Gpu_Model"] = model


    # Search for ARM Mali model
    if pd.isna(result["Gpu_Model"]):

        match = re.search(
            r"\bMali\s+(.+)$",
            text,
            re.IGNORECASE
        )

        if match:
            model = match.group(1).strip()

            result["Gpu_Model"] = model


    # Return the extracted GPU information
    return pd.Series(result)


# Apply the extraction function to the GPU column
gpu_features = gpu.apply(
    extract_gpu_info
)


# Define the new GPU columns
gpu_columns = [
    "Gpu_Type",
    "Gpu_Model"
]


# Add the extracted GPU columns to the DataFrame
df[gpu_columns] = gpu_features[gpu_columns]


# Get the current column order
columns = df.columns.tolist()


# Remove the GPU feature columns from their current positions
for column in gpu_columns:
    if column in columns:
        columns.remove(column)


# Insert Gpu_Type into the DataFrame
columns.insert(16, "Gpu_Type")

# Insert Gpu_Model into the DataFrame
columns.insert(17, "Gpu_Model")


# Apply the new column order
df = df[columns]


# Display the first 20 rows
print("\nGPU Information:")

print(
    df[
        [
            "Gpu",
            "Gpu_Type",
            "Gpu_Model"
        ]
    ].head(20)
)


# Check missing values
print("\nMissing Values:")

print(
    df[
        [
            "Gpu_Type",
            "Gpu_Model"
        ]
    ].isnull().sum()
)


# Display GPUs where the model was not extracted
print("\nGPU Model Missing:")

print(
    df[
        df["Gpu_Model"].isnull()
    ][
        [
            "Gpu",
            "Gpu_Type",
            "Gpu_Model"
        ]
    ].head(30)
)


# Display unique GPU types
print("\nUnique Values - GPU Type:")

print(
    df["Gpu_Type"]
    .value_counts(dropna=False)
)


# Display unique GPU models
print("\nUnique Values - GPU Model:")

print(
    df["Gpu_Model"]
    .value_counts(dropna=False)
)


# Display 30 random GPU samples
print("\nRandom GPU Samples:")

print(
    df[
        [
            "Gpu",
            "Gpu_Type",
            "Gpu_Model"
        ]
    ]
    .sample(30, random_state=42)
    .to_string(index=False)
)


# Display GPUs where the type was not extracted
print("\nGPU Type Missing:")

print(
    df[
        df["Gpu_Type"].isna()
    ]["Gpu"]
    .unique()
)
```

```python
# Creating a function to plot a bar chart

def bar_plot(column_name, title, xlabel, ylabel):

    # Set the style and color codes for seaborn
    sns.set(style="whitegrid", color_codes=True)

    # figsize
    plt.figure(figsize=[20, 10], dpi=300)

    # Obtaining Unique Values and Their Counts
    unique_counts = df[column_name].value_counts()[:]

    # barplot
    if column_name == "Inches":
        ax = sns.barplot(x=unique_counts.index, y=unique_counts.values, palette="pastel", edgecolor="black", linewidth=3)
    elif column_name == "Cpu_GHz":
        ax = sns.barplot(x=unique_counts.index, y=unique_counts.values, palette="pastel", edgecolor="black", linewidth=3)
    elif column_name == "Ram_GB":
        ax = sns.barplot(x=unique_counts.index, y=unique_counts.values, palette="pastel", edgecolor="black", linewidth=3)
    elif column_name == "Memory_Capacity_GB_1":
        ax = sns.barplot(x=unique_counts.index, y=unique_counts.values, palette="pastel", edgecolor="black", linewidth=3)
    elif column_name == "Storage_Count":
        ax = sns.barplot(x=unique_counts.index, y=unique_counts.values, palette="pastel", edgecolor="black", linewidth=3)
    elif column_name == "Memory_Capacity_GB_2":
        ax = sns.barplot(x=unique_counts.index, y=unique_counts.values, palette="pastel", edgecolor="black", linewidth=3)
    elif column_name == "Storage_Type_2":
       ax = sns.barplot(x=unique_counts.index, y=unique_counts.values, palette="pastel", edgecolor="black", linewidth=3) 
    else:
        ax = sns.barplot(x=unique_counts.values, y=unique_counts.index, palette="pastel", edgecolor="black", linewidth=3)

    # axis
    for container in ax.containers:

        ax.bar_label(container, fontsize=15, padding=5)

    ax.set_title(title, fontsize=28, weight="bold", x=0.5, y=1.03)
    ax.set_xlabel(xlabel, fontsize=20, weight="bold")
    ax.set_ylabel(ylabel, fontsize=20, weight="bold")
    ax.tick_params(axis="both", labelsize=18)

    plt.show()
```

```python
# Calculate the first quartile (Q1), the third quartile (Q3), and the interquartile range (IQR) for the laptop prices.
Q1 = df["Price_euros"].quantile(0.25)
Q3 = df["Price_euros"].quantile(0.75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
lower_bound = Q1 - 1.5 * IQR
Upper_bound = Q3 + 1.5 * IQR

print("lower_bound:", lower_bound)
print("Upper_bound:", Upper_bound)
```

```python
# Identify medical charge values above the upper IQR boundary and count the potential outliers.

outlier = df[df["Price_euros"] > Upper_bound]

print("Number of outliers:", len(outlier))
```

```python
def pie_plot(column_name, title):


    label = df[column_name].unique()
    count = df[column_name].value_counts()

    # pie chart
    fig = px.pie(values=count, names=label, hole=0.4)

    # update layout
    fig.update_layout(
        width=1400,
        height=800,
        font_size=20,
        font_weight="bold",
        template="plotly_dark",

        # title
        title=title,
        title_font_size=25,

    )

    # update traces
    fig.update_traces(
        marker=dict(line=dict(color="#000000", width=4))
    )
    

    fig.show()
```
